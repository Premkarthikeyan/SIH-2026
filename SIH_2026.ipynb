```python
# STEP 4: Prepare features and predefined splits

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score, classification_report,
    confusion_matrix, f1_score
)
from sklearn.ensemble import RandomForestClassifier

# Exclude identifiers, target, split, and possible leakage columns
exclude_cols = [
    "pair_id",
    "entity_id",
    "is_match",
    "split",
    "scenario",
    "mismatched_fields"
]

X = df.drop(columns=exclude_cols, errors="ignore")
y = df["is_match"].astype(int)

# Use the dataset's predefined splits
train_mask = df["split"].eq("train")
val_mask = df["split"].eq("val")
test_mask = df["split"].eq("test")

X_train, y_train = X[train_mask], y[train_mask]
X_val, y_val = X[val_mask], y[val_mask]
X_test, y_test = X[test_mask], y[test_mask]

# Detect numeric and categorical columns
numeric_cols = X_train.select_dtypes(include="number").columns.tolist()
categorical_cols = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Impute missing values and encode categories
preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), numeric_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_cols)
])

# Train a baseline model
model = Pipeline([
    ("preprocess", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

model.fit(X_train, y_train)

# Evaluate on validation data first
val_pred = model.predict(X_val)

print("Validation accuracy:", round(accuracy_score(y_val, val_pred), 4))
print("Validation macro F1:", round(f1_score(
    y_val, val_pred, average="macro"
), 4))
print("\nValidation classification report:")
print(classification_report(y_val, val_pred, zero_division=0))
print("\nValidation confusion matrix:")
print(confusion_matrix(y_val, val_pred))
```

```python
import os
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Reload dataset if the runtime lost its variables
file_path = "/content/ml2_cross_document_verification.csv"

if "df" not in globals() or not isinstance(df, pd.DataFrame):
    df = pd.read_csv(file_path)

# Use engineered comparison features only (defined outside the IF block)
feature_cols = [
    "ocr_confidence_a", "ocr_confidence_b",
    "name_ratio", "name_token_sort_ratio",
    "name_token_set_ratio", "name_initial_match",
    "dob_available", "dob_exact_match",
    "dob_diff_days", "dob_year_match",
    "reg_available", "reg_norm_exact_match",
    "reg_similarity", "address_available",
    "address_similarity", "pincode_match"
]
feature_cols = [c for c in feature_cols if c in df.columns]

train = df[df["split"] == "train"]
val = df[df["split"] == "val"]
X_train = train[feature_cols].apply(pd.to_numeric, errors="coerce")
X_val = val[feature_cols].apply(pd.to_numeric, errors="coerce")
y_train = train["is_match"].astype(int)
y_val = val["is_match"].astype(int)
medians = X_train.median().fillna(0)
X_train = X_train.fillna(medians)
X_val = X_val.fillna(medians)

model_simple = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)
model_simple.fit(X_train, y_train)
pred = model_simple.predict(X_val)

print("Features used:", feature_cols)
print("Validation accuracy:", round(accuracy_score(y_val, pred), 4))
print("Validation macro F1:", round(f1_score(y_val, pred, average="macro"), 4))
print(classification_report(y_val, pred, zero_division=0))
print("\nFeature importance:")
display(
    pd.DataFrame({
        "feature": feature_cols,
        "importance": model_simple.feature_importances_
    }).sort_values("importance", ascending=False)
)
```

```python

# STEP 8: Final evaluation on the held-out test split

test = df[df["split"] == "test"]

X_test = test[feature_cols].apply(pd.to_numeric, errors="coerce")
X_test = X_test.fillna(medians)
y_test = test["is_match"].astype(int)

test_pred = model_simple.predict(X_test)

print("TEST ACCURACY:",
      round(accuracy_score(y_test, test_pred), 4))
print("TEST MACRO F1:",
            round(f1_score(y_test, test_pred, average="macro"), 4))
print("\nTest classification report:")
print(classification_report(
                y_test, test_pred, zero_division=0
))

print("\nTest confusion matrix:")
print(confusion_matrix(y_test, test_pred))
```

```python

# SIH 2026: Complete preprocessing pipeline

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# 1. Define target and remove columns that could leak the answer
target = "case_label"

leakage_columns = [
    "is_consistent",
        "has_discrepancy",
            "has_missing_fields",
                "has_expired_document",
                    "mismatch_fields",
                        "missing_fields",
                            "expiry_status_a",
                                "case_id",
                                    "faculty_id"
]

                                    # Use the original dataframe to build the inputs
X = df.drop(
                                        columns=[target] + [
                                                col for col in leakage_columns if col in df.columns
                                                    ]
                                                    ).copy()

y = df[target].copy()

                                                    # 2. Convert date columns into numeric timestamps
                                                    # This lets the numerical preprocessing pipeline handle them.
date_cols = X.select_dtypes(
    include=["datetime", "datetimetz"]
                                                        ).columns.tolist()

for col in date_cols:
                                                            dates = pd.to_datetime(X[col], errors="coerce")
X[col] = dates.map(
                                                                        lambda value: value.timestamp()
                                                                                if pd.notna(value) else np.nan
                                                                                    )

                                                                                    # 3. Split the data before fitting preprocessing steps
X_train, X_test, y_train, y_test = train_test_split(
                                                                                        X,
                                                                                            y,
                                                                                                test_size=0.20,
                                                                                                    random_state=42,
                                                                                                        stratify=y
                                                                                                        )

                                                                                                        # 4. Identify numerical and categorical columns
num_cols = X_train.select_dtypes(
                                                                                                            include=np.number
                                                                                                            ).columns.tolist()

cat_cols = X_train.select_dtypes(
                                                                                                                exclude=np.number
                                                                                                                ).columns.tolist()

print("Numerical columns:", len(num_cols))
print("Categorical columns:", len(cat_cols))

                                                                                                                # 5. Preprocess numerical columns
num_pipeline = Pipeline([
                                                                                                                    ("imputer", SimpleImputer(strategy="median")),
                                                                                                                        ("scaler", StandardScaler())
                                                                                                                        ])

                                                                                                                        # 6. Preprocess categorical columns
cat_pipeline = Pipeline([
                                                                                                                            ("imputer", SimpleImputer(strategy="most_frequent")),
                                                                                                                                ("encoder", OneHotEncoder(handle_unknown="ignore"))
                                                                                                                                ])

                                                                                                                                # 7. Combine both pipel
preprocessor = ColumnTransformer([
                                                                                                                                    ("num", num_pipeline, num_cols),
                                                                                                                                        ("cat", cat_pipeline, cat_cols)
                                                                                                                                        ])

                                                                                                                                        # 8. Fit only on training data and transform bo
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

                                                                                                                                        # 9. Display results
print("\nOriginal dataset:", df.shape)
print("Training input:", X_train_processed.shape)
print("Testing input:", X_test_processed.shape)
print("Training output:", y_train.shape)
print("Testing output:", y_test.shape)
print("\nTraining class distribution:")
print(y_train.value_counts())
print("\nPreprocessing completed!")
```

```python
import pandas as pd
import numpy as np
from catboost import CatBoostClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

# 1. Copy the original training and testing data
X_train_cb = X_train.copy()
X_test_cb = X_test.copy()

# 2. Identify datetime columns and convert them to numeric (timestamps) so CatBoost can handle them
date_cols = X_train_cb.select_dtypes(include=["datetime", "datetime64", "datetime64[ns]"]).columns.tolist()
# Also check if any datetime columns are treated as objects
for col in X_train_cb.columns:
    if col not in date_cols:
        try:
            # Try converting a sample to see if it's datetime-like
            pd.to_datetime(X_train_cb[col].dropna().iloc[:3])
            # If successful and it's not a categorical or simple ID
            if "date" in col.lower() or "time" in col.lower():
                date_cols.append(col)
        except:
            pass

for col in set(date_cols):
    # Train
    dates_train = pd.to_datetime(X_train_cb[col], errors="coerce")
    X_train_cb[col] = dates_train.map(lambda val: val.timestamp() if pd.notna(val) else np.nan)
    # Test
    dates_test = pd.to_datetime(X_test_cb[col], errors="coerce")
    X_test_cb[col] = dates_test.map(lambda val: val.timestamp() if pd.notna(val) else np.nan)

# 3. Identify categorical columns
cat_features = X_train_cb.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

for col in cat_features:
    X_train_cb[col] = X_train_cb[col].fillna("__MISSING__").astype(str)
    X_test_cb[col] = X_test_cb[col].fillna("__MISSING__").astype(str)

# 4. Train CatBoost
cb_model = CatBoostClassifier(
    iterations=500,
    depth=5,
    learning_rate=0.05,
    loss_function="MultiClass",
    auto_class_weights="Balanced",
    random_seed=42,
    verbose=100,
    allow_writing_files=False
)

cb_model.fit(
    X_train_cb,
    y_train,
    cat_features=cat_features
)

cb_pred = cb_model.predict(X_test_cb).ravel()

print("\nCatBoost Accuracy:", round(accuracy_score(y_test, cb_pred), 4))
print("Macro F1-score:", round(f1_score(y_test, cb_pred, average="macro"), 4))

print("\nClassification Report:")
print(classification_report(y_test, cb_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, cb_pred, labels=["consistent", "discrepant", "missing_fields"]))
```
